# Para que se usa T-Student ?

En una situacion ideal, tú conocerías la desviación estándar poblacional ($\sigma$). Pero en la vida real, si estás tomando una muestra para adivinar el promedio de una población, es ilógico pensar que vas a conocer la dispersión de esa misma población. ¡Si no conoces el promedio, menos vas a conocer su dispersión!Como no conoces $\sigma$, tienes que estimarla usando la desviación de tu muestra ($s$).El problema es que al estimar, estás agregando más incertidumbre. Ya no solo estás adivinando la media ($\mu$), sino que también estás adivinando la dispersión ($\sigma$). Para compensar esta "doble ignorancia" (especialmente si tu muestra es pequeña, menos de 30 datos), la estadística te obliga a usar la distribución $t$.

#Intervalos de confianza para la proporcion

## Que es la proporcion ?

Una proporción es la fracción, parte o porcentaje de un grupo total que cumple con una característica específica.Mientras que la media poblacional ($\mu$) te responde a la pregunta "¿Cuánto es el promedio?", la proporción ($p$) te responde a la pregunta "¿Cuántos (o qué porcentaje) son de este tipo?".

##Que vendria siendo entonces intervalos de confianza para la proporcion ?

Los intervalos de confianza para la proporción son un rango de porcentajes que dice que la condición que encontré en mi muestra, se va a repetir en un cierto pedazo (porcentaje) de mi población total.

# Formula para sacar los intervalos de confianza para la proporcion

image.png

## Desglose de la formula

La fórmula se divide en un límite inferior (resta) y un límite superior (suma). Consta de las siguientes partes:

* **$\hat{p}$ (Proporción Muestral):** Es el porcentaje exacto de "éxitos" (casos que cumplen la condición) que encontramos dentro de nuestra muestra. Es el centro exacto de nuestro intervalo.
  * **¿Cómo se calcula?** Se divide la cantidad de éxitos ($x$) por el tamaño total de la muestra ($n$). Su fórmula es: $\hat{p} = \frac{x}{n}$

* **$1 - \hat{p}$ (Proporción de Fracasos):** Es simplemente el porcentaje restante de tu muestra. Si tu proporción de éxito fue del 30% ($0.30$), tu proporción de fracaso es del 70% ($0.70$).

* **$n$ (Tamaño de la Muestra):** Es la cantidad total de datos, personas o elementos que evaluaste en tu grupo de estudio.

* **$z_{\frac{\alpha}{2}}$ (Valor Crítico Z):** Es el multiplicador estadístico que define nuestro Nivel de Confianza (ej. 95%). El subíndice $\frac{\alpha}{2}$ indica que nuestro riesgo de error total ($\alpha$) se ha partido en dos partes iguales, colocando una mitad en la cola izquierda y otra en la cola derecha de la distribución normal.

* **$\sqrt{\frac{\hat{p} \cdot (1 - \hat{p})}{n}}$ (Error Estándar de la Proporción):** Toda esta raíz cuadrada es el equivalente a la desviación estándar, pero calculada específicamente para porcentajes. Nos dice qué tanto varían los datos basándose directamente en la misma proporción.

* **El bloque $z_{\frac{\alpha}{2}} \cdot \sqrt{\frac{\hat{p} \cdot (1 - \hat{p})}{n}}$ (Margen de Error):** Es el resultado final que le vamos a restar y sumar a nuestro $\hat{p}$ central para construir los límites del intervalo.

---

## 2. Requisitos de Aplicación (Aproximación a la Normal)


Esta fórmula matemática funciona gracias a un "truco" estadístico: asume que, si tenemos suficientes datos, una distribución de proporciones (que originalmente es Binomial) adoptará la forma perfecta de una Campana de Gauss (Distribución Normal).

Para que esta suposición matemática sea válida y nuestro intervalo no entregue resultados mentirosos o sesgados, se **deben cumplir estrictamente estas dos reglas** antes de aplicar la fórmula:

1. **$n \cdot \hat{p} \ge 10$** *(El número de éxitos observados en tu muestra debe ser mayor o igual a 10).*
2. **$n \cdot (1 - \hat{p}) \ge 10$** *(El número de fracasos observados en tu muestra debe ser mayor o igual a 10).*

**¿Qué pasa si no se cumplen?** Significa que tu muestra es demasiado pequeña o tu característica es demasiado rara. El intervalo pierde fiabilidad y tendrías que usar métodos estadísticos más avanzados (métodos exactos) en lugar de esta fórmula general.

#Codigo para sacar los intervalos para la proporcion

```python
import numpy as np
import pandas as pd
import scipy.stats as st

'''
Datos y cálculos necesarios para el Intervalo de Proporción:
- Tamaño Muestral (n)
- Contabilizar éxitos (x)
- Proporción muestral (p_hat)
- Nivel de significancia (alpha) y Valor crítico Z
- Error estándar (ee)
'''

# 1.- Lectura de la muestra (ejemplo con Excel)
df = pd.read_excel("Nombre_del_archivo.xlsx")

# 2.- Extracción de columna de la tabla y la guardamos en una variable
muestra = df["Nombre_columna"]

# 3.- Contabilizamos el tamaño de los datos (n)
# CORRECCIÓN: Es len(), no np.len()
n = len(muestra)

# 4.- Extraemos los "éxitos" (filtramos los datos que cumplen la condición)
exitos = muestra[muestra == "atributo_de_la_columna"]

# 5.- Contabilizamos los éxitos (x)
num_exitos = len(exitos)

# 6.- Cálculo de proporción muestral (p_hat)
proporcion_muestral = num_exitos / n

# 7.- Sacar valor crítico Z
# Si queremos un 95% de confianza, nuestro alpha (margen de error) es 0.05
alpha = 0.05

z = st.norm.ppf(1 - alpha/2)

# 8.- Calculamos el error estándar
# Aplicamos la fórmula de la raíz cuadrada: sqrt( (p * (1-p)) / n )
ee = np.sqrt(proporcion_muestral * (1 - proporcion_muestral) / n)

# 9.- Cálculo de los intervalos de confianza
limite_inferior = proporcion_muestral - (z * ee)
limite_superior = proporcion_muestral + (z * ee)

ic = (limite_inferior, limite_superior)

# 10.- Imprimir los resultados (El paso que faltaba)
print(f"Límite inferior: {ic[0] * 100:.2f}%")
print(f"Límite superior: {ic[1] * 100:.2f}%")
print(f"Conclusión: Estamos un {(1-alpha)*100:.0f}% seguros de que la proporción real de la población está entre {ic[0] * 100:.2f}% y {ic[1] * 100:.2f}%.")
```

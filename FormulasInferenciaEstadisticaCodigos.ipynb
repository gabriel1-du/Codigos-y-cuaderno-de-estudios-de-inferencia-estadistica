# Codigo para calcular intervalos de confianza para la media

```python
# Librerías
import numpy as np
import pandas as pd
import scipy.stats as st

'''
Criterios:
- Desviación estándar poblacional (sigma) ES CONOCIDA o muestra grande: Distribución Normal (Z).
- Desviación estándar poblacional ES DESCONOCIDA: t-Student.

Cosas que necesitamos calcular o definir:
- Tamaño de los datos (n)
- Media muestral (x̄)
- Desviación estándar poblacional (sigma) -> Dato dado
- Error estándar (sigma / √n)
'''

# 1. Lectura de la tabla y extracción de la muestra
tabla = pd.read_excel('tabla.xlsx')
muestra = tabla['Nombre_Columna']

# 2. Variables que guardan los datos iniciales
# IMPORTANTE: Aquí debes ingresar el valor real de tu desviación poblacional
desviacion_poblacional_conocida = 15.0

# 3. Estadísticos necesarios
tamaño_muestral = len(muestra)
media_muestral = np.mean(muestra)

# 4. Cálculo del error estándar
error_estandar = desviacion_poblacional_conocida / np.sqrt(tamaño_muestral)

# 5. Cálculo del intervalo de confianza (Distribución Normal)
intervalo_confianza = st.norm.interval(confidence=0.95, loc=media_muestral, scale=error_estandar
)

# 6. Impresión de resultados
# Al imprimir tendremos un arreglo con el límite inferior y superior
print(f"Límite inferior: {intervalo_confianza[0]:.2f} , Límite superior: {intervalo_confianza[1]:.2f}")
```

#Codigo para caluclar los intervalos de confianza para la proporcion

```python
import numpy as np
import pandas as pd
import scipy.stats as st

'''
Datos y cálculos necesarios para el Intervalo de Proporción:
- Tamaño Muestral (n)
- Contabilizar éxitos (x)
- Proporción muestral (p_hat)
- Nivel de significancia (alpha) y Valor crítico Z
- Error estándar (ee)
'''

# 1.- Lectura de la muestra (ejemplo con Excel)
df = pd.read_excel("Nombre_del_archivo.xlsx")

# 2.- Extracción de columna de la tabla y la guardamos en una variable
muestra = df["Nombre_columna"]

# 3.- Contabilizamos el tamaño de los datos (n)
# CORRECCIÓN: Es len(), no np.len()
n = len(muestra)

# 4.- Extraemos los "éxitos" (filtramos los datos que cumplen la condición)
exitos = muestra[muestra == "atributo_de_la_columna"]

# 5.- Contabilizamos los éxitos (x)
num_exitos = len(exitos)

# 6.- Cálculo de proporción muestral (p_hat)
proporcion_muestral = num_exitos / n

# 7.- Sacar valor crítico Z
# Si queremos un 95% de confianza, nuestro alpha (margen de error) es 0.05
alpha = 0.05

z = st.norm.ppf(1 - alpha/2)

# 8.- Calculamos el error estándar
# Aplicamos la fórmula de la raíz cuadrada: sqrt( (p * (1-p)) / n )
ee = np.sqrt(proporcion_muestral * (1 - proporcion_muestral) / n)

# 9.- Cálculo de los intervalos de confianza
limite_inferior = proporcion_muestral - (z * ee)
limite_superior = proporcion_muestral + (z * ee)

ic = (limite_inferior, limite_superior)

# 10.- Imprimir los resultados (El paso que faltaba)
print(f"Límite inferior: {ic[0] * 100:.2f}%")
print(f"Límite superior: {ic[1] * 100:.2f}%")
print(f"Conclusión: Estamos un {(1-alpha)*100:.0f}% seguros de que la proporción real de la población está entre {ic[0] * 100:.2f}% y {ic[1] * 100:.2f}%.")
```
